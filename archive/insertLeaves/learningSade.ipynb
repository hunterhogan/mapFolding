```python
from __future__ import annotations

def build_albums(n_max):
    """
    Sade's 'albums' construction (Sur les Chevauchements des Permutations, secs. 4-7).

    A permutation P of {1,...,n} (fixed to start with 1, per Sade's circular-invariance
    remark in sec.2) is 'sans chevauchement' (overlap-free / planar) iff no two edges
    (a,a+1) and (b,b+1), with a,b of the SAME parity, cross when drawn as arcs over the
    linear order of positions in P. ('Cross' = exactly one endpoint of one edge's
    position-interval lies strictly inside the other's -- i.e. interleaved, not nested,
    not disjoint.)

    Returns: dict n -> list of overlap-free permutations of degree n (as tuples),
             dict (n,i) -> count  == Sade's V(n,i) == A259689 T(n,i)
    """
    def crosses(iv1, iv2):
        (x1, x2), (y1, y2) = iv1, iv2
        if x1 > x2: x1, x2 = x2, x1
        if y1 > y2: y1, y2 = y2, y1
        return (x1 < y1 < x2 < y2) or (y1 < x1 < y2 < x2)

    albums = {2: [(1, 2)]}
    V = {}

    for n in range(2, n_max):
        current = albums[n]
        next_album = []
        counts_this_row = []
        for P in current:
            pos = {v: idx for idx, v in enumerate(P)}
            valid_gaps = 0
            for g in range(1, n + 1):  # insertion gaps 1..n (never before the leading 1)
                Pp = P[:g] + (n + 1,) + P[g:]
                pos_p = pos.copy()
                pos_p[n + 1] = g
                for k in range(g + 1, n + 1):
                    pos_p[Pp[k]] = k

                new_edge = (pos_p[n], pos_p[n + 1])
                ok = True
                a = n - 2
                while a >= 1:
                    if crosses(new_edge, (pos_p[a], pos_p[a + 1])):
                        ok = False
                        break
                    a -= 2
                if ok:
                    valid_gaps += 1
                    next_album.append(Pp)
            counts_this_row.append(valid_gaps)
        albums[n + 1] = next_album
        for i in counts_this_row:
            V[n, i] = V.get((n, i), 0) + 1

    return albums, V

albums, V = build_albums(14)

for n in range(12, 14):
    r = max((i for (nn, i) in V if nn == n), default=0)
    row = [V.get((n, i), 0) for i in range(2, r + 1)]
    print(n, row, "sum =", sum(row), " P_n =", len(albums[n]))

print(albums[14][-1])
```
